```python
nlp = spacy.load("en_core_web_sm")

print("spaCy English model loaded successfully!")
```

```python
text = """Thiss PHONE is Amazng!!! I buyed it last wek, and the battry life is sooo

goood but the camra quality isnt that greaat."""

print("Raw Customer Review:")
print(text)
```

```python
lower_text = text.lower()

print("Lowercase Text:")
print(lower_text)
```

```python
no_punctuation = re.sub(r'[^\w\s]', '', lower_text)

print("Text after removing punctuation:")
print(no_punctuation)
```

```python
normalized_text = re.sub(r'\s+', ' ', no_punctuation).strip()

print("Final Normalized Text:")
print(normalized_text)
```

```python
import nltk

nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize

nltk_tokens = word_tokenize(normalized_text)

print("NLTK Tokens:")
print(nltk_tokens)
```

```python
doc = nlp(normalized_text)

spacy_tokens = [token.text for token in doc]

print("spaCy Tokens:")
print(spacy_tokens)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

nltk_filtered = [word for word in nltk_tokens if word not in stop_words]

print("NLTK Tokens After Stop-word Removal:")
print(nltk_filtered)
```

```python
doc = nlp(normalized_text)

print("spaCy document created successfully!")
```

```python
spacy_filtered = [token.text for token in doc if not token.is_stop]

print("spaCy Tokens After Stop-word Removal:")
print(spacy_filtered)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in nltk_filtered]

print("Stemmed Words:")
print(stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

nltk_lemmas = [lemmatizer.lemmatize(word) for word in nltk_filtered]

print("NLTK Lemmatized Words:")
print(nltk_lemmas)
```

```python
spacy_lemmas = [token.lemma_ for token in doc if not token.is_stop]

print("spaCy Lemmatized Words:")
print(spacy_lemmas)
```

```python
misspelled_word = "battry"
correct_word = "battery"

print("Misspelled Word:", misspelled_word)
print("Correct Word:", correct_word)
```

```python
from nltk.metrics.distance import edit_distance
```

```python
distance_cost_1 = edit_distance(
    misspelled_word,
    correct_word,
    substitution_cost=1
)

print("Edit Distance (Substitution Cost = 1):", distance_cost_1)
```

```python
distance_cost_2 = edit_distance(
    misspelled_word,
    correct_word,
    substitution_cost=2
)

print("Edit Distance (Substitution Cost = 2):", distance_cost_2)
```

```python
print("----- EDIT DISTANCE RESULTS -----")

print("Misspelled Word:", misspelled_word)
print("Correct Word:", correct_word)

print("Edit Distance (Substitution Cost = 1):", distance_cost_1)
print("Edit Distance (Substitution Cost = 2):", distance_cost_2)
```
