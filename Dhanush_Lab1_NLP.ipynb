## Part A: Theoretical Analysis

### 1. Morphology
The word "arrived" can be divided into:
arrive + ed.

"arrive" is the root word and "-ed" is the suffix.
The suffix "-ed" indicates past tense.

### 2. Lexical Analysis
- driver → Noun
- arrived → Verb

### 3. Syntax
The sentence "The driver arrived on time." can be divided into:

- Noun Phrase (NP): "The driver"
- Verb Phrase (VP): "arrived on time"

### 4. Semantics
The literal meaning of "ride" is a journey or trip taken in a vehicle.

### 5. Pragmatics
If "The driver arrived on time" is written in a 1-star review,
the literal statement is a fact, but depending on the surrounding
context it could potentially be interpreted sarcastically.

```python
import re
```

```python
text = """UNBELIEVABLY... the driver, Alex, arrivED on 12/05/2024!
He was super helpful. However, the ride cost $45.50 which is pricey.
Contact support@taxi-tech.com or call (555) 019-2834 for refunds.
I was unhappy with the wait time though."""
```

```python
print(text)
```

```python
normalized_text = text.lower()

print(normalized_text)
```

```python
# Task 1: Text Normalization and Cleaning

# Convert text to lowercase
normalized_text = text.lower()

# Remove unwanted punctuation while keeping $, ., (, )
cleaned_text = re.sub(r"[^a-z0-9\s$.()]", "", normalized_text)

# Replace multiple spaces with a single space
cleaned_text = re.sub(r"\s+", " ", cleaned_text)

# Remove spaces from beginning and end
cleaned_text = cleaned_text.strip()

print("Normalized and Cleaned Text:")
print(cleaned_text)
```

```python
date_pattern = r"\d{2}/\d{2}/\d{4}"

date = re.findall(date_pattern, text)

print("Date found:", date)
```

```python
price_pattern = r"\$\d+\.\d+"

price = re.findall(price_pattern, text)

print("Price found:", price)
```

```python
email_pattern = r"[\w.-]+@[\w.-]+\.\w+"

emails = re.findall(email_pattern, text)

print("Email found:", emails)
```

```python
phone_pattern = r"\(\d{3}\)\s\d{3}-\d{4}"

phones = re.findall(phone_pattern, text)

print("Phone found:", phones)
```

```python
print("----- Information Extraction -----")

print("Date:", date)
print("Price:", price)
print("Email:", emails)
print("Phone:", phones)
```
